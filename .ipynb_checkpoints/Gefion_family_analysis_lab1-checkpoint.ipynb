# Обработка и визуализация данных астероидного семейства Гефьён

**Цель работы:** построение гистограмм основных физических характеристик определённого семейства астероидов

**Используемые дата сеты:** Nesvorny HCM asteroids families и JPL Small-Body Query

**Инструменты:** Python (pandas, matplotlib, seaborn, Jupyter notebook)

**Исследуемое семейство астероидов:** Gefion

**Ссылка на HCM asteroid families:** https://sbn.psi.edu/pds/resource/nesvornyfam.html

**Ссылка на JPL Small-Body Query:** https://ssd.jpl.nasa.gov/tools/sbdb_query.html

## Ход работы

### 1. Выполним импорт необходимых библиотек 

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

### 2. Выгрузим данные по семейству *Gefion* из данных *Nesvorny HCM Asteroid Families* за 2015 год
(https://sbnarchive.psi.edu/pds4/non_mission/ast.nesvorny.families_V2_0/data/families_2015/516_gefion.tab)<br>
Создамим DataFrame из полученных данных и найдём разброс значений больших полуосей и эксцентриситетов астероидов

```python
hcm_af_table = pd.read_table("516_gefion.tab.txt", sep='\\s+',
                            names = ['id', 'a', 'e', 'i', 'm', 'c_param', 'fam_num', 'par_num', 'name'])
a_max = hcm_af_table.iloc[:, 1].max()
e_max = hcm_af_table.iloc[:, 2].max()

print(f'Максимальные значения большой полуоси и эксцентриситета: {a_max}, {e_max}')

a_min = hcm_af_table.iloc[:, 1].min()
e_min = hcm_af_table.iloc[:, 2].min()

print(f'Минимальные значения большой полуоси и эксцентриситета: {a_min}, {e_min}')
```

### 3. Создадим функцию для обработки значений ID из базы *JPL Small-Body Query* 
Убираем префикс, начинающийся на 20...0

```python
def cut_id(id):
    id_str = str(id)[1:]
    while id_str[0] == '0':
        id_str = id_str[1:]
    return int(id_str)
```

### 4. Выгрузим следующие данные по астероидам: SPK-ID, prim design, H, diameter, albedo, rot_per, Spec_B, Spec_T, a, e, i, q, B-V, U-B, I-R. 
В выборку попадут только те тела, у которых значения большой полуоси и эксцентриситета в пределах, вычисленных в пункте 2. <br>
После внесём данные из *JPL Small-Body Query* в DataFrame, проведём некоторые преобразования и объединим с предыдущей таблицей в DataFrame *result*

```python
jpl_sbq_table = pd.read_csv("sbdb_query_results.csv")
jpl_sbq_table.iloc[:, 0] = jpl_sbq_table.iloc[:, 0].map(cut_id) # Удаление префиксов
hcm_af_without_double = hcm_af_table.drop(columns=['a', 'e', 'i'], errors='ignore') # Удаление дубликатов столбцов из HCM Asteroid Families 
jpl_sbq_table.rename(columns={'spkid': 'id'}, inplace=True)
result = pd.merge(hcm_af_without_double, jpl_sbq_table, on='id')
result.to_csv('result_lab_1.csv')
```

### 5. Сделаем выборку для нахождения данных по родительскому телу

```python
par_id = result['par_num'].iloc[0]
par = result.query(f'id == {par_id}')
```

### 6. Выгрузим данные по первым 500 астероидам из базы *JPL Small-Body Query* и вычислим средние значения

```python
fst_500 = pd.read_csv("fst_500.csv")

avg_albedo = np.mean(fst_500['albedo'])
avg_a = np.mean(fst_500['a'])
avg_e = np.mean(fst_500['e'])
avg_i = np.mean(fst_500['i'])
```

### 7. Получим выборки по значению диаметра: больше 10 км, от 2 до 10 км, меньше 2 км
Сделаем допущение и скажем, что астероиды с диаметром меньше 2 км не имеют установленных значений 

```python
over_10km = result[result['diameter'] > 10]
under_10km  = result[(result['diameter'] > 2) & (result['diameter'] < 10)]
under_2km = result[result['diameter'].isna()]
```

```python
list_of_columns = list(result.columns)[-6:] + ['H', 'rot_per'] # Внесение названий столбцов в отдельный список
list_of_dfs = [result, over_10km, under_10km, under_2km]
```

### 8. Построим график распределения известных диаметров астероидов

```python
plt.figure(figsize=(12, 6))
sns.histplot(data=result, x='diameter', element="step")
plt.axvline(x=par['diameter'][1], color='red', label= f'({par_id}) Гефьён')
plt.yscale('log')
plt.xlabel('Диаметр, км')
plt.ylabel('Количество астероидов')
plt.title('Распределение диаметра астероидов')
plt.legend(loc='upper right')

plt.show()
```

### 9. Построим графики по параметрам Spec_B, Spec_T, a, e, i, q, H, rot_per для всего семейства, для астероидов семейства с диаметром больше 10 км, от 2 км до 10 км и до 2 км

#### 9.1 Гистограммы для каждого диапазона диаметров и для всего семейства на разных координатных плоскостях

```python
fig, ax = plt.subplots(nrows = 8, ncols = 4, figsize=(15, 20))

for i, axis in enumerate(ax.flat):
    j = i % 4
    k = i // 4
    sns.histplot(data = list_of_dfs[j], x = list_of_columns[k], element="step", ax = axis)
    axis.axvline(x=par[list_of_columns[k]][1], color='red', label= f'({par_id}) Гефьён')
    if list_of_columns[k] == 'a':
        axis.axvline(x=2.71, color='purple', label='Резонанс 8:3')
        axis.axvline(x=2.82, color='purple', label='Резонанс 5:2')
    axis.legend(loc='upper right', framealpha=0)
    axis.set_xlabel('')
    axis.grid(True, alpha=0.5)
    
    match i:
        case 0:
            axis.set_title('Всё семейство')
        case 1:
            axis.set_title('> 10 км')
        case 2:
            axis.set_title('[2; 10] км')
        case 3:
            axis.set_title('< 2 км')
            
    if not(j):
        axis.set_ylabel(list_of_columns[k], fontsize=11)
    else:
        axis.set_ylabel('')

plt.tight_layout()
plt.show()
```

#### 9.2 Гистограммы  для каждого диапазона диаметров и для всего семейства на одних координатных плоскостях

```python
fig, ax = plt.subplots(nrows = 8, ncols = 1, figsize=(8, 30))

for i, axis in enumerate(ax.flat):
    sns.histplot(data = list_of_dfs[0], x = list_of_columns[i], element = "step", ax = axis, label = 'Всё семейство')
    sns.histplot(data = list_of_dfs[1], x = list_of_columns[i], element = "step", ax = axis, label = '> 10 км')
    sns.histplot(data = list_of_dfs[2], x = list_of_columns[i], element = "step", ax = axis, label = '[2; 10] км')
    sns.histplot(data = list_of_dfs[3], x = list_of_columns[i], element = "step", ax = axis, label = '< 2 км')
    axis.axvline(x = par[list_of_columns[i]][1], color='red', label= f'({par_id}) Гефьён')
    if list_of_columns[i] == 'a':
        axis.axvline(x=2.71, color='purple', label='Резонанс 8:3')
        axis.axvline(x=2.82, color='purple', label='Резонанс 5:2')
    axis.legend(loc='upper right', framealpha = 0)
    axis.set_xlabel('')
    axis.grid(True, alpha=0.5)        
    axis.set_ylabel(list_of_columns[i], fontsize=11)

plt.tight_layout()
plt.show()
```

### 10. Построим график распределения альбедо астероидов и сравним с первыми 500

```python
plt.figure(figsize=(14, 6))
sns.histplot(data=result, x='albedo', element="step", label='Семейство Гефьёна')
sns.histplot(data=fst_500, x='albedo', element="step", label='Первые 500 астероидов')
plt.axvline(x=par['albedo'][1], color='red', label= f'({par_id}) Гефьён')
plt.axvline(x=avg_albedo, color='green', label='Среднее 500 астероидов')
plt.yscale('log')
plt.xlabel('Альбедо')
plt.ylabel('Количество астероидов')
plt.title('Распределение альбедо астероидов')
plt.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

### 11. Построим график распределения больших полуосей астероидов и сравним с первыми 500

```python
plt.figure(figsize=(12, 6))
sns.histplot(data=result, x='a', element="step", label='Семейство Гефьёна')
sns.histplot(data=fst_500, x='a', element="step", label='Первые 500 астероидов')
plt.axvline(x=par['a'][1], color='red', label= f'({par_id}) Гефьён')
plt.axvline(x=avg_a, color='green', label='Среднее 500 астероидов')

plt.axvline(x=2.71, color='purple', label='Орбитальные резонансы')
plt.axvline(x=1.9, color='purple')
plt.axvline(x=2.06, color='purple')
plt.axvline(x=2.25, color='purple')
plt.axvline(x=2.33, color='purple')
plt.axvline(x=2.5, color='purple')
plt.axvline(x=2.82, color='purple')
plt.axvline(x=2.95, color='purple')
plt.axvline(x=3.03, color='purple')
plt.axvline(x=3.075, color='purple')
plt.axvline(x=3.27, color='purple')
plt.axvline(x=3.47, color='purple')
plt.axvline(x=3.7, color='purple')

plt.yscale('log')
plt.xlabel('Большая полуось, а. е.')
plt.ylabel('Количество астероидов')
plt.title('Распределение больших полуосей астероидов')
plt.legend(loc='upper right')
plt.show()
```

### 12. Построим график распределения эксцентриситета астероидов и сравним с первыми 500

```python
plt.figure(figsize=(12, 6))
sns.histplot(data=result, x='e', element="step", label='Семейство Гефьёна')
sns.histplot(data=fst_500, x='e', element="step", label='Первые 500 астероидов')
plt.axvline(x=par['e'][1], color='red', label= f'({par_id}) Гефьён')
plt.axvline(x=avg_e, color='green', label='Среднее 500 астероидов')
plt.yscale('log')
plt.xlabel('Эксцентриситет')
plt.ylabel('Количество астероидов')
plt.title('Распределение эксцентриситета астероидов')
plt.legend(loc='upper right')
plt.show()
```

### 13. Построим график распределения углов наклонения орбиты астероидов и сравним с первыми 500

```python
plt.figure(figsize=(12, 6))
sns.histplot(data=result, x='i', element="step", label='Семейство Гефьёна')
sns.histplot(data=fst_500, x='i', element="step", label='Первые 500 астероидов')
plt.axvline(x=par['i'][1], color='red', label= f'({par_id}) Гефьён')
plt.axvline(x=avg_i, color='green', label='Среднее 500 астероидов')
plt.yscale('log')
plt.xlabel('Угол наклонения орбиты, °')
plt.ylabel('Количество астероидов')
plt.title('Распределение наклонений астероидов')
plt.legend(loc='upper right')
plt.show()
```

### 14. Построим графики по параметрам H, diameter, albedo, rot_per

```python
sns.pairplot(result, vars=['H', 'diameter', 'albedo', 'rot_per'], hue='spec_B')
plt.show()
```
